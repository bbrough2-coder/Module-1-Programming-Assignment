### 3.1 How many seconds are in an hour?

There are 60 seconds in a minute and 60 minutes in an hour. 

```python
60 * 60
```

### 3.2 Assign the result to a variable

I assigned the number of seconds in an hour (3600) to a variable called 'seconds_per_hour'.

```python
seconds_per_hour = 60 * 60
```

### 3.3 How many seconds are in a day?

There are 24 hours in a day. I used my 'seconds_per_hour' variable to calculate the total number of seconds.

```python
seconds_per_hour * 24
```

### 3.4 Save the result in a variable

I saved the number of seconds in a day to a variable called 'seconds_per_day'.

```python
seconds_per_day = seconds_per_hour * 24
```

### 3.5 Divide using floating-point division

I divided 'seconds_per_day' by 'seconds_per_hour' using regular division (/). The result is a floating-point number.

```python
seconds_per_day / seconds_per_hour
```

### 3.6 Divide using integer division

I divided 'seconds_per_day' by 'seconds_per_hour' using integer division (//). The result agrees with the floating-point value from the previous question, except it does not include the decimal point and zero. 

```python
seconds_per_day // seconds_per_hour
```

**Answer** Yes, the result agrees with the floating-point answer from 3.5. Both results are 24, but the floating-point answer displays as 24.0.
